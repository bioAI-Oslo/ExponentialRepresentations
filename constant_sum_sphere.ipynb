```python
import numpy as np
from scipy.linalg import expm

# Step 1: Dimensions
n = 5  # example: 5D space
one_vec = np.ones((n,))/np.sqrt(n)

# Step 2: Create R ∈ R^{(n-1) × n}, with rowspace orthogonal to 1
# This is done by orthonormalizing a basis of the orthogonal complement of 1
# Step 2: Perform Gram-Schmidt starting from the ones vector

# Now R is the projection matrix onto the plane orthogonal to 1
R = orthogonal_basis(one_vec)
R = R[1:]


# Step 3: Create random skew-symmetric A ∈ R^{(n-1) × (n-1)}
def random_skew_symmetric(dim):
    M = np.random.randn(dim, dim)
    return M - M.T

A = random_skew_symmetric(n - 1)

# # Step 4: Compute Q = R^T e^A R
# Q = R.T @ expm(A) @ R

# # Step 5: Test zᵀ Q x = 0 when z = 1
# x = np.random.randn(n)
# z = np.ones(n)

# result = z.T @ Q @ x
# print("zᵀ Q x =", result)

# # # Should be approximately 0 (up to numerical precision)
# assert np.allclose(result, 0, atol=1e-12), "Condition not satisfied"
```
