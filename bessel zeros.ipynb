```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import jn_zeros

# Compute zeros of the Bessel function J_0
num_zeros = 15
zeros = jn_zeros(0, num_zeros)
```

```python
# Compute ratio of subsequent zeros
ratios = zeros[1:] / zeros[:-1]

cum_average =  np.cumsum(ratios) / np.arange(1, len(ratios) + 1)

# Compute moving average of the ratios
# Plotting
plt.figure(figsize=(6, 4))
plt.plot(ratios, 'ko-', linewidth = 0.5)
plt.plot(np.arange(len(cum_average)), cum_average, 'k--', linewidth = 1)
plt.xlabel('Bessel Zero')
plt.ylabel(r'$j_{0,n+1} / j_{0,n}$')
plt.savefig("./figures/bessel_zeros.pdf", dpi = 300, bbox_inches = "tight", transparent = True)

plt.show()
```
