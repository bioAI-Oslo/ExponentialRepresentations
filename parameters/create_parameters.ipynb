```python
import numpy as np 
import torch
import scipy
```

```python
def block_diag(fill_vals):
    # Helper function to create block-diagonal skew-symmetric matrices
    blocks = []
    for i in range(len(fill_vals)):
        block = torch.zeros((2,2))
        
        block[0,1] = fill_vals[i]
        block[1,0] = -block[0,1]
        blocks.append(block)
    block_form = torch.block_diag(*blocks)
    return block_form.to(torch.float32)
```

```python
# Roots of unity generator eigenvalues
M = 3 # exmample with M roots of unity
n_copies = 50  # to match number of neurons in Gaussian cell model, for simplicity

phi = np.arange(0, M)* np.pi/M
phi_plot = np.concatenate([phi, phi + np.pi])  # add negative rotations
phi = np.concatenate([phi for _ in range(n_copies)])

sigma_A_RoU = block_diag(np.cos(phi))
sigma_B_RoU = block_diag(np.sin(phi))

torch.save(sigma_A_RoU, "sigma_A_RoU.pt")
torch.save(sigma_B_RoU, "sigma_B_RoU.pt")
```

```python
# Gaussian generator eigenvalues
N = 2*M*n_copies # number of "neurons"

sigma_A_pc = block_diag(np.sqrt(2)*torch.randn(N//2)) 
sigma_B_pc = block_diag(np.sqrt(2)*torch.randn(N//2)) 
sigma_C_pc = block_diag(np.sqrt(2)*torch.randn(N//2))

torch.save(sigma_A_pc, "sigma_A_pc.pt")
torch.save(sigma_B_pc, "sigma_B_pc.pt")
torch.save(sigma_C_pc, "sigma_C_pc.pt")
```
